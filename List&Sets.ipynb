```python


# 1. CREATING SETS
# Note: Sets are unordered, unchangeable, and do not allow duplicate values.
fruits = {"apple", "banana", "cherry"}
numbers = set([1, 2, 3, 3, 4, 4, 5]) # Duplicates (3, 4) are automatically removed

print("--- 1. Creating Sets ---")
print("Fruits:", fruits)
print("Unique Numbers:", numbers)


```

```python
# 2. ADDING & REMOVING ELEMENTS
fruits.add("orange")                  # Add a single item
fruits.update(["mango", "grape"])     # Add multiple items

fruits.remove("banana")               # Removes 'banana' (raises KeyError if not found)
fruits.discard("pineapple")           # Removes if exists, ignores if not found (safer)

print("\n--- 2. Adding & Removing ---")
print("Updated Fruits:", fruits)


```

```python
# 3. SET OPERATIONS (Venn Diagram Math)
set_a = {1, 2, 3, 4, 5}
set_b = {4, 5, 6, 7, 8}

# Union: Items in either A or B
union_set = set_a | set_b             # or set_a.union(set_b)

# Intersection: Items in BOTH A and B
intersect_set = set_a & set_b         # or set_a.intersection(set_b)

# Difference: Items in A but NOT in B
diff_set = set_a - set_b              # or set_a.difference(set_b)

# Symmetric Difference: Items in A or B, but NOT both
sym_diff_set = set_a ^ set_b          # or set_a.symmetric_difference(set_b)

print("\n--- 3. Set Operations ---")
print("Set A:", set_a)
print("Set B:", set_b)
print("Union (A | B):", union_set)
print("Intersection (A & B):", intersect_set)
print("Difference (A - B):", diff_set)
print("Symmetric Difference (A ^ B):", sym_diff_set)


```

```python
# 4. SUBSETS & SUPERSETS
small_set = {1, 2}
big_set = {1, 2, 3, 4}

print("\n--- 4. Subsets & Supersets ---")
print("Is small_set a subset of big_set?:", small_set.issubset(big_set))    # True
print("Is big_set a superset of small_set?:", big_set.issuperset(small_set))  # True
print("Do A and B share no elements?:", set_a.isdisjoint({9, 10}))          # True


# 5. COMMON USE CASE: Removing duplicates from a list
raw_list = ["cat", "dog", "cat", "bird", "dog", "fish"]
unique_list = list(set(raw_list))

print("\n--- 5. Removing Duplicates from a List ---")
print("Original List:", raw_list)
print("Unique List:", unique_list)
```

List in Python

```python
# =======================================================
# PYTHON LISTS CHEATSHEET FOR GOOGLE COLAB
# =======================================================

# 1. CREATING LISTS
# Lists are ordered, changeable (mutable), and allow duplicate values.
fruits = ["apple", "banana", "cherry"]
numbers = [1, 2, 3, 4, 5]
mixed = [1, "hello", 3.14, True]

print("--- 1. Creating Lists ---")
print("Fruits:", fruits)
print("Mixed types:", mixed)


# 2. ACCESSING & SLICING ELEMENTS
# Python uses zero-based indexing.
first_fruit = fruits[0]        # "apple"
last_fruit = fruits[-1]        # "cherry" (negative index starts from end)
sub_list = numbers[1:4]        # [2, 3, 4] (index 1 up to, but not including, index 4)

print("\n--- 2. Accessing & Slicing ---")
print("First fruit:", first_fruit)
print("Last fruit:", last_fruit)
print("Sliced numbers (index 1 to 3):", sub_list)


# 3. ADDING & MODIFYING ELEMENTS
fruits[1] = "blackberry"       # Modify an item directly
fruits.append("orange")        # Add item to the end
fruits.insert(1, "mango")      # Insert "mango" at index 1
fruits.extend(["grape", "kiwi"]) # Add multiple items to the end

print("\n--- 3. Adding & Modifying ---")
print("Updated fruits:", fruits)


# 4. REMOVING ELEMENTS
fruits.remove("apple")         # Removes first occurrence of "apple"
popped_item = fruits.pop(2)    # Removes and returns item at index 2
del fruits[0]                  # Deletes item at index 0

print("\n--- 4. Removing Items ---")
print("Popped item:", popped_item)
print("Fruits after removals:", fruits)


# 5. SORTING & REVERSING
nums = [42, 10, 7, 99, 1]
nums.sort()                    # Sorts in-place (ascending)
nums.reverse()                 # Reverses in-place (descending)

print("\n--- 5. Sorting & Reversing ---")
print("Sorted & Reversed numbers:", nums)


# 6. LIST COMPREHENSION (Fast way to build lists)
# Syntax: [expression for item in iterable if condition]
squares = [x**2 for x in range(1, 6)]
evens = [x for x in range(10) if x % 2 == 0]

print("\n--- 6. List Comprehension ---")
print("Squares of 1-5:", squares)
print("Even numbers under 10:", evens)
```
